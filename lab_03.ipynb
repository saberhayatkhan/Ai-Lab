```python
from collections import deque
import time

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)  # 0 represents the blank tile

MOVES = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}


def actions(state):
    """Return the legal moves of the blank tile."""
    i = state.index(0)
    row, col = divmod(i, 3)
    legal = []
    if row > 0: legal.append('Up')
    if row < 2: legal.append('Down')
    if col > 0: legal.append('Left')
    if col < 2: legal.append('Right')
    return legal


def result(state, action):
    """Transition model: swap the blank with the neighbouring tile."""
    i = state.index(0)
    j = i + MOVES[action]
    s = list(state)
    s[i], s[j] = s[j], s[i]
    return tuple(s)


def goal_test(state):
    return state == GOAL


def is_solvable(state):
    tiles = [t for t in state if t != 0]
    inversions = sum(1 for a in range(len(tiles))
                     for b in range(a + 1, len(tiles)) if tiles[a] > tiles[b])
    return inversions % 2 == 0


def solve_bfs(start):
    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    while frontier:
        state, path = frontier.popleft()
        if goal_test(state):
            return path, expanded, len(explored)
        expanded += 1
        for a in actions(state):
            child = result(state, a)
            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [a]))
    return None, expanded, len(explored)


def show(state):
    for r in range(3):
        print(' '.join(str(x) if x else '_' for x in state[r*3:r*3+3]))
import random
import matplotlib
matplotlib.use("Agg")          # save to file; remove this line to show a window
import matplotlib.pyplot as plt


# ---------------- Lab Task 3 ----------------
def random_start(n_moves, seed=None):
    """Apply n random legal moves to the goal (always gives a solvable state)."""
    rng = random.Random(seed)
    state = GOAL
    prev = None
    for _ in range(n_moves):
        opts = actions(state)
        # avoid immediately undoing the previous move
        undo = {'Up': 'Down', 'Down': 'Up', 'Left': 'Right', 'Right': 'Left'}
        if prev is not None and undo[prev] in opts and len(opts) > 1:
            opts.remove(undo[prev])
        a = rng.choice(opts)
        state = result(state, a)
        prev = a
    return state


random_moves = [5, 10, 15, 20, 25]
rows = []

for k, n in enumerate(random_moves):
    s = random_start(n, seed=k + 1)
    assert is_solvable(s)
    t0 = time.perf_counter()
    path, expanded, generated = solve_bfs(s)
    t1 = time.perf_counter()
    rows.append((n, len(path), expanded, generated, (t1 - t0) * 1000))

print(f"{'Random moves':>12} | {'Depth':>5} | {'Expanded':>8} | {'Generated':>9} | {'Time (ms)':>9}")
print("-" * 56)
for n, d, e, g, t in rows:
    print(f"{n:>12} | {d:>5} | {e:>8} | {g:>9} | {t:>9.2f}")

# ---- plot depth vs nodes expanded ----
# a random walk can wander back toward the goal, so sort by true depth
by_depth = sorted(rows, key=lambda r: r[1])
depths = [r[1] for r in by_depth]
expanded_counts = [r[2] for r in by_depth]

plt.figure(figsize=(7, 4.5))
plt.plot(depths, expanded_counts, marker='o')
for d, e in zip(depths, expanded_counts):
    plt.annotate(str(e), (d, e), textcoords="offset points", xytext=(0, 8), ha='center')
plt.yscale('log')
plt.xlabel("Solution depth (moves)")
plt.ylabel("Nodes expanded (log scale)")
plt.title("8-Puzzle BFS: depth vs nodes expanded")
plt.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.savefig("depth_vs_nodes.png", dpi=150)
print("\nPlot saved as depth_vs_nodes.png")

print("""
Comment on growth:
The number of expanded nodes grows exponentially with solution depth
(roughly b^d, where b is the effective branching factor, about 1.5-2 for the
8-puzzle because the blank has 2-4 moves and already-explored states are
pruned). On the log-scale plot the curve is close to a
straight line, which confirms exponential growth. Each extra move of depth
multiplies the work by about a constant factor. Note: a random walk can
wander back toward the goal, so the true optimal depth can be smaller than
the number of random moves applied (see the 25-move row).
""")
```

```python
from collections import deque
import time

# ---------------- Post-Lab: Missionaries and Cannibals ----------------
# STATE REPRESENTATION
#   state = (M, C, B)
#   M = missionaries on the LEFT bank  (0..3)
#   C = cannibals    on the LEFT bank  (0..3)
#   B = boat position: 1 = left bank, 0 = right bank
#   Right bank counts are implied: (3 - M, 3 - C)
#
# INITIAL STATE : (3, 3, 1)   everyone and the boat on the left
# GOAL STATE    : (0, 0, 0)   everyone and the boat on the right
#
# ACTIONS: the boat carries (m, c) people, with 1 <= m + c <= 2
BOAT_LOADS = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

START = (3, 3, 1)
GOAL = (0, 0, 0)


def is_safe(M, C):
    """A state is safe if on neither bank cannibals outnumber missionaries
    (a bank with 0 missionaries is always safe)."""
    if not (0 <= M <= 3 and 0 <= C <= 3):
        return False
    Mr, Cr = 3 - M, 3 - C
    if M > 0 and C > M:
        return False
    if Mr > 0 and Cr > Mr:
        return False
    return True


def actions(state):
    """Return the legal boat loads (m, c) from this state."""
    M, C, B = state
    legal = []
    for m, c in BOAT_LOADS:
        if B == 1:                       # boat on left: people leave the left bank
            newM, newC = M - m, C - c
        else:                            # boat on right: people return to the left
            newM, newC = M + m, C + c
        if is_safe(newM, newC):
            legal.append((m, c))
    return legal


def result(state, action):
    """Transition model: move the boat and the chosen people to the other bank."""
    M, C, B = state
    m, c = action
    if B == 1:
        return (M - m, C - c, 0)
    return (M + m, C + c, 1)


def goal_test(state):
    return state == GOAL


def solve_bfs(start):
    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    while frontier:
        state, path = frontier.popleft()
        if goal_test(state):
            return path, expanded, len(explored)
        expanded += 1
        for a in actions(state):
            child = result(state, a)
            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [a]))
    return None, expanded, len(explored)


def describe(action, boat_on_left):
    m, c = action
    direction = "left -> right" if boat_on_left else "right -> left"
    return f"{m} missionary(ies) + {c} cannibal(s)  [{direction}]"


def print_solution(start, path):
    state = start
    print(f"Start : left=(M{state[0]},C{state[1]})  right=(M{3-state[0]},C{3-state[1]})  boat=left")
    for step, a in enumerate(path, 1):
        boat_left = state[2] == 1
        state = result(state, a)
        side = "left" if state[2] == 1 else "right"
        print(f"Step {step}: {describe(a, boat_left)}")
        print(f"        left=(M{state[0]},C{state[1]})  right=(M{3-state[0]},C{3-state[1]})  boat={side}")


t0 = time.perf_counter()
path, expanded, generated = solve_bfs(START)
t1 = time.perf_counter()

print_solution(START, path)
print("\nSolution length:", len(path))
print("Nodes expanded :", expanded)
print("Nodes generated:", generated)
print(f"Time taken     : {(t1 - t0) * 1000:.2f} ms")
```
